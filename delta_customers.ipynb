```python
%sql
-- Завдання 1. Довідник клієнтів у форматі Delta
DROP TABLE IF EXISTS workspace.default.customers_delta;

CREATE TABLE workspace.default.customers_delta
USING DELTA
AS 
SELECT  
    c_custkey,
    c_name,
    c_nationkey,
    c_mktsegment,
    c_acctbal
FROM samples.tpch.customer
WHERE
    c_custkey BETWEEN 1 AND 20;


SELECT * FROM customers_delta;
```

```python
%sql
DESCRIBE DETAIL workspace.default.customers_delta;
```

```python
%sql

DESCRIBE HISTORY workspace.default.customers_delta;
```

```python
%sql
-- Завдання 2. Зміни в довіднику; додано клієнтів з c_custkey від 21 до 25 включно

INSERT INTO workspace.default.customers_delta
SELECT  
    c_custkey,
    c_name,
    c_nationkey,
    c_mktsegment,
    c_acctbal
FROM samples.tpch.customer
WHERE
    c_custkey BETWEEN 21 AND 25;
```

```python
%sql
-- Завдання 2. Зміни в довіднику; клієнт з c_custkey, що дорівнює 1, має сегмент AUTOMOBILE;

UPDATE workspace.default.customers_delta
SET c_mktsegment = 'AUTOMOBILE'
WHERE c_custkey = 1;
```

```python
%sql
-- Завдання 2. Зміни в довіднику; клієнтів з відʼємним балансом c_acctbal у таблиці немає.

DELETE FROM workspace.default.customers_delta
WHERE c_acctbal < 0;
```

```python
%sql 
SELECT *
FROM workspace.default.customers_delta
ORDER BY c_custkey
```

```python
%sql
DESCRIBE HISTORY customers_delta;
```

```python
%sql
-- Завдання 3. Пакет змін із CRM

MERGE INTO workspace.default.customers_delta AS t
USING( 
    SELECT
    c_custkey,
    c_name,
    c_nationkey,
    c_mktsegment,
    c_acctbal + 100 AS c_acctbal
    FROM
    samples.tpch.customer
    WHERE
        c_custkey BETWEEN 21 AND 30
    ) AS s
ON t.c_custkey = s.c_custkey
WHEN MATCHED THEN UPDATE SET *
WHEN NOT MATCHED THEN INSERT *
;
```

```python
%sql
SELECT *
FROM workspace.default.customers_delta
ORDER BY c_custkey;

```

```python
# Завдання 4. Історія версій

versions = (
    spark
    .sql("""DESCRIBE HISTORY workspace.default.customers_delta""")
    .select(
        "version",
        "operation",
        "operationMetrics"
    )
    .orderBy("version", ascending=True)
    )
display(versions)
```

```python
%sql
-- Завдання 5. Time travel. Завдання 5.1. Клієнти і баланс за версіями

SELECT 
    0 AS version,
    COUNT(*) AS customers_cnt,
    SUM(c_acctbal) AS total_acctbal
FROM workspace.default.customers_delta
VERSION AS OF 0

UNION ALL

SELECT 
    1 AS version,
    COUNT(*) AS customers_cnt,
    SUM(c_acctbal) AS total_acctbal
FROM workspace.default.customers_delta
VERSION AS OF 1

UNION ALL

SELECT 
    2 AS version,
    COUNT(*) AS customers_cnt,
    SUM(c_acctbal) AS total_acctbal
FROM workspace.default.customers_delta
VERSION AS OF 2

UNION ALL

SELECT 
    3 AS version,
    COUNT(*) AS customers_cnt,
    SUM(c_acctbal) AS total_acctbal
FROM workspace.default.customers_delta
VERSION AS OF 3

UNION ALL

SELECT 
    4 AS version,
    COUNT(*) AS customers_cnt,
    SUM(c_acctbal) AS total_acctbal
FROM workspace.default.customers_delta
VERSION AS OF 4

ORDER BY version;
```

```python
# Завдання 5.2. Стан на момент видалення

timestamp_for_delete = (
    spark
    .sql("""DESCRIBE HISTORY customers_delta;""")
    .filter("operation = 'DELETE' ")
    .select("timestamp")
    .first()["timestamp"] 
)

table = spark.sql(f"""
          SELECT
          TIMESTAMP '{timestamp_for_delete}' AS as_of,
          COUNT(*) AS customers_cnt        
          FROM workspace.default.customers_delta
          TIMESTAMP AS OF '{timestamp_for_delete}'
          """)

display(table)
```

```python
%sql
-- Завдання 6. Помилка і відновлення

UPDATE workspace.default.customers_delta
SET c_acctbal = 0;

SELECT * 
FROM workspace.default.customers_delta;
```

```python
%sql
-- Restored table
RESTORE workspace.default.customers_delta TO VERSION AS OF 4;

SELECT * 
FROM workspace.default.customers_delta;
```

```python
%sql
-- Завдання 6.2. Перевірка повернення

SELECT
    (SELECT COUNT(*)
        FROM (
            SELECT *
            FROM workspace.default.customers_delta
            
            EXCEPT

            SELECT *
            FROM workspace.default.customers_delta
            VERSION AS OF 4
        ) AS diff
    ) AS only_current,


    (SELECT COUNT(*)
        FROM (
           SELECT *
           FROM workspace.default.customers_delta
            VERSION AS OF 4

            EXCEPT

            SELECT *
            FROM workspace.default.customers_delta
        ) AS merge_diff
    ) AS only_merge_version
```
